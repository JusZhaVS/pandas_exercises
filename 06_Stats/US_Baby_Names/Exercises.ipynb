### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/US_Baby_Names/US_Baby_Names_right.csv). 

### Step 3. Assign it to a variable called baby_names.

```python
url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/US_Baby_Names/US_Baby_Names_right.csv'
df = pd.read_csv(url)
df.head()
```

### Step 4. See the first 10 entries

```python
df.head(10)
```

### Step 5. Delete the column 'Unnamed: 0' and 'Id'

```python
df.drop(columns=['Unnamed: 0', 'Id'], inplace=True)
df.head()
```

### What year has the highest number of baby names in the dataset?

```python
int(df.groupby('Year').sum().idxmax()['Name']) #just want the one for name
```

### Step 6. Is there more male or female names in the dataset?

```python
num_fem = int(df['Gender'].value_counts()['F'])
num_male = int(df['Gender'].value_counts()['M'])
if num_fem >= num_male:
    print("More females")
else:
    print("More males")
```

### Step 7. Group the dataset by name and assign to names

```python
names = df.groupby('Name', as_index=True)[['Count']].sum() #to get it as a dataframe, added the extra brackets around 'Count'

names.sort_values('Count', ascending=False).head() #if you used a series, wouldn't need the 'Count' because would be autosorted
```

### Step 8. How many different names exist in the dataset?

```python
len(names) #all names unique alr since we grouped by it, so just the length
```

### Step 9. What is the name with most occurrences?

```python
#names.groupby('Name').sum().idxmax()['Count'] don't need another groupBy cuz alr grouped by it
names['Count'].idxmax()
```

### Step 10. How many different names have the least occurrences?

```python
names.head()
```

```python
#print((names[names['Count'] == names['Count'].min()]).sum()) this one is wrong, actually adds those counts
print(len(names[names['Count'] == names['Count'].min()]))


(names['Count'] == names['Count'].min()).sum() #this one correct, actually sums the bools
```

### Step 11. What is the median name occurrence?

```python
names['Count'].median()
```

### Step 12. What is the standard deviation of names?

```python
int(names['Count'].std())
```

### Step 13. Get a summary with the mean, min, max, std and quartiles.

```python
names.describe()
```
