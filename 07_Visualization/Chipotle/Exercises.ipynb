### Step 5. Create a histogram of the top 5 items bought

```python
top5 = chipo.groupby('item_name').agg({
    'quantity': 'sum'
}).sort_values('quantity', ascending=False).iloc[:5]

print(top5.head())

top5.plot(kind='bar')
plt.title('Top 5 Items Bought')
plt.xlabel('Item')
plt.ylabel('Total Quantity Bought')
#plt.xticks(rotation=45, ha='right')
plt.show()
```
