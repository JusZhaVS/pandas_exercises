# Apple Stock

### Introduction:

We are going to use Apple's stock price.


### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/09_Time_Series/Apple_Stock/appl_1980_2014.csv)

```python
url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/09_Time_Series/Apple_Stock/appl_1980_2014.csv'
df = pd.read_csv(url)
```

### Step 3. Assign it to a variable apple

```python
apple = df
apple.head()
```

### Step 4.  Check out the type of the columns

```python
apple.dtypes
```

### Step 5. Transform the Date column as a datetime type

```python
apple['Date'] = pd.to_datetime(apple['Date'])
```

```python
apple.dtypes
```

### Step 6.  Set the date as the index

```python
apple.set_index('Date', inplace=True)
```

### Step 7.  Is there any duplicate dates?

```python
apple.index.is_unique
```

### Step 8.  Ops...it seems the index is from the most recent date. Make the first entry the oldest date.

```python
apple.sort_index(ascending=True, inplace=True)
apple.head()
```

### Step 9. Get the last business day of each month

```python
apple_last_business_day = apple.resample('BME').last() #takes the last row in each of these months, which represents the last business day
apple_last_business_day.head()
```

### Step 10.  What is the difference in days between the first day and the oldest

```python
#question is a bit unclear, basically asking for difference between earliest and latest day

(apple.index.max() - apple.index.min()).days #.days pulls out just the integer count, so the time difference which is like a timestamp value is converted to just the number itself
```

### Step 11.  How many months in the data we have?

```python
apple_months = apple.resample('BME')
len(apple_months)
```

```python
apple_months = apple.resample('BME')
```

### Step 12. Plot the 'Adj Close' value. Set the size of the figure to 13.5 x 9 inches

```python
apple['Adj Close'].plot(figsize=(13.5, 9))

plt.xlabel('Date')
plt.ylabel('Closing Price')
plt.title('Relation between Closing Price and Date')

plt.show()
```
